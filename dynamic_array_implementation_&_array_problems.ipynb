```python
import ctypes

class DynamicArray:

  def __init__(self):
    self.size = 0
    self.capacity = 1
    self.arr_A = self.make_array(self.capacity)
    self.array = self.arr_A

  def make_array(self,capacity):
    return (capacity * ctypes.py_object)()

  def len(self):
    return self.size

  def get_item(self,k):
    if not 0<=k<self.size:
      raise IndexError("Index out of range")
    return self.arr_A[k]

  def append(self,item):
    if self.size == self.capacity:
      self.resize(2*self.capacity)

    self.arr_A[self.size] = item
    self.size = self.size + 1



  def resize(self,new_capacity):
    arr_B = self.make_array(new_capacity)

    for i in range(self.size):
      arr_B[i] = self.arr_A[i]
    self.arr_A = arr_B
    self.capacity = new_capacity

  def insert(self,index,item):
    if not 0<=index<=self.size:
      raise("Index out of range error : ")

    if self.size == self.capacity:
      self.resize(2*self.capacity)

    for i in range(self.size,index,-1):
      self.arr_A[i] = self.arr_A[i-1]

    self.arr_A[index] = item
    self.size = self.size + 1

  def delete(self,index):
    if not 0<=index<self.size:
      raise("Index out of range error : ")

    for i in range(index,self.size-1):
      self.arr_A[i] = self.arr_A[i+1]

    self.arr_A[self.size-1] = 0
    self.size = self.size - 1

    if 0 <= self.size < self.capacity//4:
      self.resize(self.capacity//2)






```

```python
# Reverse an array

def reverse_array(array):
  return array[::-1]

reverse_array([1,2,3])
```

```python
# Find the missing number in a given range [0, n]

def missing_numbers(array):

  max_num = 1
  min_num = 0

  for i in array:
    if i>max_num:
      max_num = i

    if i<min_num:
      min_num = i
  dup = []
  for j in range(min_num, max_num+1):
    dup.append(j)
  res = set(dup) - set(array)
  res_sort = sorted(list(res))
  return res_sort

missing_numbers([-2,9,7,4])
```

```python
# Move all zeros to the end of the array (in-place)

def moving_zeros(array):


  b = []
  for i in array:
    if i == 0:
      b.append(i)
      array.remove(i)
  res = array + b
  return res

moving_zeros([23,0,45,0,231,55,8,0,99])

```

```python
# Find the intersection of two arrays

def array_intersection(arr1, arr2):
  return list(set(arr1).intersection(set(arr2)))

array_intersection([1,2,2,4],[1,2,3])
```
